```python
import csv
import json
import re
import time
from datetime import datetime, timezone
from pathlib import Path
from urllib.parse import urlencode

import requests
from bs4 import BeautifulSoup

# NJIT Engage / CampusLabs public event search API.
API_BASE = "https://njit.campuslabs.com/engage/api/discovery/event/search"
EVENT_PAGE_BASE = "https://njit.campuslabs.com/engage/event"
# Confirmed from NJIT Engage category facets: 20748 == Volunteering.
VOLUNTEERING_CATEGORY_ID = "20748"

HEADERS = {
    "Accept": "application/json",
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
    ),
}

OUTPUT_JSON = Path("njit_volunteering_events.json")
OUTPUT_CSV = Path("njit_volunteering_events_app.csv")


def clean_html(html: str | None) -> str:
    """Convert CampusLabs HTML descriptions into readable plain text."""
    if not html:
        return ""
    soup = BeautifulSoup(html, "html.parser")
    text = soup.get_text(" ", strip=True)
    return re.sub(r"\s+", " ", text).strip()


def event_url(event_id: str | int | None) -> str | None:
    if not event_id:
        return None
    return f"{EVENT_PAGE_BASE}/{event_id}"


def fetch_volunteering_events(take: int = 100, max_pages: int = 20) -> list[dict]:
    """Fetch upcoming NJIT Engage events whose category is Volunteering."""
    events = []
    seen_ids = set()
    skip = 0
    now_utc = datetime.now(timezone.utc)
    ends_after = now_utc.isoformat()

    for page in range(max_pages):
        params = {
            "take": take,
            "skip": skip,
            "categoryIds": VOLUNTEERING_CATEGORY_ID,
            "endsAfter": ends_after,
            "forceOrderBy": "true",
            "orderByField": "StartsOn",
            "orderByDirection": "ascending",
        }
        url = f"{API_BASE}?{urlencode(params)}"
        response = requests.get(url, headers=HEADERS, timeout=20)
        response.raise_for_status()
        data = response.json()
        items = data.get("value", [])

        print(f"Fetched page {page + 1}: {len(items)} events")
        if not items:
            break

        for item in items:
            event_id = str(item.get("id", "")).strip()
            if not event_id or event_id in seen_ids:
                continue
            seen_ids.add(event_id)
            events.append(item)

        if len(items) < take:
            break

        skip += take
        time.sleep(0.25)

    return events


def parse_event_datetime(value: str | None) -> datetime | None:
    if not value:
        return None
    return datetime.fromisoformat(value.replace("Z", "+00:00"))


def is_current_or_upcoming(item: dict, now_utc: datetime | None = None) -> bool:
    """Keep approved events that have not ended yet."""
    if item.get("status") != "Approved":
        return False

    now_utc = now_utc or datetime.now(timezone.utc)
    ends_on = parse_event_datetime(item.get("endsOn"))
    starts_on = parse_event_datetime(item.get("startsOn"))
    if ends_on is not None:
        return ends_on >= now_utc
    return starts_on is not None and starts_on >= now_utc


def normalize_event(item: dict) -> dict:
    """Keep the app-facing fields we care about."""
    event_id = str(item.get("id", "")).strip()
    description = clean_html(item.get("description"))
    return {
        "source": "njit_engage",
        "event_id": event_id,
        "title": item.get("name") or "",
        "organization_name": item.get("organizationName") or "",
        "organization_id": item.get("organizationId"),
        "location": item.get("location") or "",
        "starts_on": item.get("startsOn"),
        "ends_on": item.get("endsOn"),
        "description": description,
        "event_url": event_url(event_id),
    }


def save_events(events: list[dict]) -> None:
    OUTPUT_JSON.write_text(json.dumps(events, indent=2, ensure_ascii=False), encoding="utf-8")

    fieldnames = [
        "source",
        "event_id",
        "title",
        "organization_name",
        "organization_id",
        "location",
        "starts_on",
        "ends_on",
        "description",
        "event_url",
    ]

    with OUTPUT_CSV.open("w", newline="", encoding="utf-8") as file:
        writer = csv.DictWriter(file, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(events)


raw_events = fetch_volunteering_events()
volunteering_events = [
    normalize_event(event)
    for event in raw_events
    if is_current_or_upcoming(event)
]
save_events(volunteering_events)

print(f"\nSaved {len(volunteering_events)} volunteering events to {OUTPUT_JSON} and {OUTPUT_CSV}")

# Preview what the app can consume.
volunteering_events[:5]
```
